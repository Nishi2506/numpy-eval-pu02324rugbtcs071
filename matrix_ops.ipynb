## Task A - Matrix Generation

```python
import numpy as np

np.random.seed(42)

A = np.random.randint(1, 11, size=(4, 3))
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix A", A.shape)
print(A)
print()
print("Matrix B", B.shape)
print(B)
```

## Task B - Dimension Validation & Dot Product

```python
# inner dims have to match for the dot product to even be defined
assert A.shape[1] == B.shape[0]

C = A.dot(B)

print("Matrix C", C.shape)
print(C)
assert C.shape == (4, 5)
```

## Task C - Reductions & Summary Statistics

```python
total = C.sum()
row_sums = C.sum(axis=1)
col_sums = C.sum(axis=0)

print("Total sum:", total)
print("Row sums:", row_sums)
print("Column sums:", col_sums)
```
